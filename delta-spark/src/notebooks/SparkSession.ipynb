```python
from delta import *
import pyspark

class Spark:
    def __init__(self):
        self.spark = Spark.get_builder()

    def get_builder():
        builder = pyspark.sql.SparkSession.builder.appName("MyApp-Delta-Spark") \
            .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension") \
            .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog") \
            .enableHiveSupport() \

        return configure_spark_with_delta_pip(builder).getOrCreate()
    
spark = Spark.get_builder()
```
